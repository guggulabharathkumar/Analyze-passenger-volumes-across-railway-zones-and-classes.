<h1 style="color:black;">
Q1: Calculate Total Passengers
</h1>

<p><b>Bloom's Level: BL-3 (Apply)</b></p>

<p>
<b>Objective:</b> Calculate the total number of passengers present in the
railway passenger dataset using Pandas and NumPy.
</p>

<p>
The Pandas <code>sum()</code> function and NumPy <code>sum()</code> function
are used to calculate the total passenger volume.
</p>

```python
# Calculate total passengers using Pandas

total_passengers = df["Passengers"].sum()

print("Total Passengers using Pandas:",
      int(total_passengers))
```

```python
# Calculate total passengers using NumPy

passenger_values = df["Passengers"].dropna().to_numpy()

total_passengers_numpy = np.sum(passenger_values)

print("Total Passengers using NumPy:",
      int(total_passengers_numpy))
```

<p>
<b>Explanation:</b>
The <code>df["Passengers"]</code> statement selects the passenger column.
The <code>sum()</code> function adds all passenger values. NumPy's
<code>np.sum()</code> performs the same operation on a NumPy array.
</p>

<h1 style="color:black;">
Q2: Filter Data by Class and Zone
</h1>

<p><b>Bloom's Level: BL-3 (Apply)</b></p>

<p>
<b>Objective:</b> Filter railway passenger records according to passenger
class and railway zone.
</p>

<p>
Pandas Boolean indexing is used to select records satisfying specific
conditions.
</p>

```python
# Display unique passenger classes

print("Available Passenger Classes:")
print(df["Class"].unique())
```

```python
# Display unique railway zones

print("Available Railway Zones:")
print(df["Zone"].unique())
```

```python
# Filter records for AC First Class

ac_first_class = df[
    df["Class"] == "AC First Class"
]

print("AC First Class Records:")
ac_first_class.head()
```

```python
# Filter records for Northern Zone

northern_zone = df[
    df["Zone"] == "Northern"
]

print("Northern Zone Records:")
northern_zone.head()
```

```python
# Filter records for Northern Zone and AC First Class

filtered_data = df[
    (df["Zone"] == "Northern") &
    (df["Class"] == "AC First Class")
]

filtered_data
```

```python
# Calculate passengers for the selected zone and class

filtered_total = filtered_data["Passengers"].sum()

print("Total passengers in Northern Zone - AC First Class:",
      int(filtered_total))
```

<p>
<b>Explanation:</b>
The <code>==</code> operator checks whether a value matches the given
condition. The <code>&amp;</code> operator combines two conditions using
AND. Therefore, only records belonging to the Northern Zone and AC First
Class are selected.
</p>

<h1 style="color:black;">
Q3: Handle Missing Passenger Counts
</h1>

<p><b>Bloom's Level: BL-4 (Analyze)</b></p>

<p>
<b>Objective:</b> Identify missing passenger counts and handle them using
an appropriate data-cleaning technique.
</p>

<p>
The missing passenger values are replaced using the median passenger count.
Median imputation is suitable for numerical data because it is less affected
by extreme values than the mean.
</p>

```python
# Count missing passenger values

missing_before = df["Passengers"].isnull().sum()

print("Missing passenger counts before handling:",
      missing_before)
```

```python
# Display records where passenger count is missing

df[df["Passengers"].isnull()]
```

```python
# Calculate median passenger count

median_passengers = df["Passengers"].median()

print("Median Passenger Count:",
      int(median_passengers))
```

```python
# Replace missing passenger values with the median

df["Passengers"] = df["Passengers"].fillna(
    median_passengers
)

print("Missing passenger values have been replaced")
```

```python
# Check missing values after handling

missing_after = df["Passengers"].isnull().sum()

print("Missing passenger counts after handling:",
      missing_after)
```

<p>
The missing passenger counts were identified using
<code>isnull()</code>. The median passenger count was calculated using
<code>median()</code>. The <code>fillna()</code> method was then used to
replace the missing values. After imputation, the number of missing passenger
values became zero.
</p>

<h1 style="color:black;">
Q4: Group Passengers by Zone and Year
</h1>

<p><b>Bloom's Level: BL-4 (Analyze)</b></p>

<p>
<b>Objective:</b> Analyze passenger volumes across railway zones and years
using Pandas grouping operations.
</p>

<p>
The <code>groupby()</code> function is used to calculate passenger totals
for different railway zones and years.
</p>

```python
# Calculate total passengers for each railway zone

zone_passengers = df.groupby(
    "Zone"
)["Passengers"].sum()

print("Passenger Volume by Railway Zone:")
print(zone_passengers)
```

```python
# Calculate total passengers for each year

year_passengers = df.groupby(
    "Year"
)["Passengers"].sum()

print("Passenger Volume by Year:")
print(year_passengers)
```

```python
# Group passengers by railway zone and year

zone_year_passengers = df.groupby(
    ["Zone", "Year"]
)["Passengers"].sum()

print("Passenger Volume by Zone and Year:")
print(zone_year_passengers)
```

```python
# Convert grouped result into a DataFrame

zone_year_table = df.groupby(["Zone", "Year"])["Passengers"].sum().reset_index()

zone_year_table.head(20)
```

```python
# Display the top 10 zone-year combinations

top_zone_year = zone_year_table.sort_values(
    "Passengers",
    ascending=False
).head(10)

top_zone_year
```

<p>
The <code>groupby()</code> function groups records according to one or more
columns. Here, the data is grouped using <code>Zone</code> and
<code>Year</code>, and the passenger counts are added using
<code>sum()</code>. This helps compare passenger volumes across railway
zones and years.
</p>

<h1 style="color:black;">
Q5: Visualize Passenger Volume Trends
</h1>

<p><b>Bloom's Level: BL-5 (Evaluate)</b></p>

<p>
<b>Objective:</b> Visualize passenger volume trends using appropriate
graphical techniques.
</p>

<p>
Bar charts are used to compare passenger volumes between railway zones,
while line charts are used to observe changes in passenger volumes across
years.
</p>

```python
# Visualize passenger volume by railway zone

plt.figure(figsize=(10, 6))

zone_passengers.sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Total Passenger Volume by Railway Zone")
plt.xlabel("Railway Zone")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Visualize passenger volume trend by year

plt.figure(figsize=(10, 6))

year_passengers.sort_index().plot(
    kind="line",
    marker="o"
)

plt.title("Passenger Volume Trend by Year")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Prepare data for zone-wise trend analysis

trend_data = df.groupby(
    ["Year", "Zone"]
)["Passengers"].sum().reset_index()

plt.figure(figsize=(12, 7))

sns.lineplot(
    data=trend_data,
    x="Year",
    y="Passengers",
    hue="Zone",
    marker="o"
)

plt.title("Passenger Volume Trends Across Railway Zones")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend(title="Railway Zone")

plt.tight_layout()
plt.show()
```

```python
# Calculate passenger volume by class

class_passengers = df.groupby(
    "Class"
)["Passengers"].sum()

plt.figure(figsize=(10, 6))

class_passengers.sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Passenger Volume by Railway Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Display the top 10 records based on passenger volume

top_10 = df.nlargest(
    10,
    "Passengers"
)

top_10[
    ["Year", "Zone", "Class", "Passengers"]
]
```
